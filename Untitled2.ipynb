# **What is Linear Regression?**

Linear Regression is a supervised learning algorithm used to predict a continuous value.

It finds a linear relationship between:

Independent variable(s) (X)

Dependent variable (y)

🔹 Basic Idea

It fits a straight line to the data.

Equation of Simple Linear Regression:

y=mx+c

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
```

```python
dataset = pd.read_csv("package.csv")
```

```python
dataset.head(3)
```

Check that this datatset follows linearlity

```python
plt.figure(figsize=(5,3))
sns.scatterplot(x="cgpa",y="package",data=dataset)
```

```python
dataset.isnull().sum()
```

```python
x1 = dataset["cgpa"]
x1.ndim
```

#it is neccesar that our dataset should present in 2d so we convert this dataset in 2D

```python
x = dataset[["cgpa"]]
y = dataset["package"]
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y , test_size = 0.2, random_state= 42)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
lr.fit(x_train, y_train)
```

```python
#y = m*x+c
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
#y = 0.57425647*x-1.0270069374542108
```

finding the accureacy

```python
lr.score(x_test, y_test)*100
```

```python
#input = 6.89
#ouput = 3.26
```

```python
lr.predict([[6.89]])
```

```python
0.57425647*6.89-1.0270069374542108
```

```python
plt.figure(figsize=(5,4))
sns.scatterplot(x="cgpa",y="package",data=dataset)
plt.plot(dataset["cgpa"] , lr.predict(x), color= "red")
plt.legend(["original line" , "predicted line"])
plt.show()
```
