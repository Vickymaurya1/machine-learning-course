📘 FEATURE SCALING – FULL NOTES
🔹 1. Definition

Feature Scaling is the process of bringing all features to a similar scale so that no feature dominates others.

🔹 2. Why Feature Scaling is Needed

👉 Example:

Feature	Value
Age	25
Salary	50,000

⚠️ Problem:

Salary is much larger → dominates model
Model gives more importance to salary ❌
🔹 3. When is Feature Scaling Required?

✅ Required for:

KNN
K-Means
SVM
Logistic Regression
Gradient Descent-based models

❌ Not required for:

Decision Trees
Random Forest

🔹 5. Difference: Standardization vs Normalization
Feature	Standardization	Normalization
Range	Not fixed	0 to 1
Outliers	Handles better	Sensitive
Use case	Most ML models	Deep learning, image data
🔹 6. Important Rule (VERY IMPORTANT 🔥)

👉 Always do:

scaler.fit(X_train)
scaler.transform(X_test)

❌ Never fit on test data (data leakage)
