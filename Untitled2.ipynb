ONE HOT ENCODING – FULL NOTES
🔹 1. Definition

One Hot Encoding (OHE) is a technique used to convert categorical data into numerical binary format (0 & 1) so that machine learning models can process it.

👉 Each category becomes a separate column.

🔹 2. Why One Hot Encoding is Needed

Machine learning models:

❌ Cannot understand text (e.g., "Red", "Blue")
❌ Misinterpret numeric labels (Label Encoding creates false order)

👉 Example problem:

Color	Label Encoding
Red	1
Blue	2
Green	3

⚠️ Model thinks: Green > Blue > Red (wrong)

🔹 3. How One Hot Encoding Works

👉 Rule:
Only one column = 1 (hot), others = 0

🎯 Example

Original:

Color
Red
Blue
Green

After OHE:

Red	Blue	Green
1	0	0
0	1	0
0	0	1
🔹 4. Key Terminology
Categorical Variable → Data with categories (Color, City)
Binary Vector → Row of 0s and 1s
Dummy Variables → New columns created after encoding
🔹 5. Implementation Methods
✅ (A) Using pandas get_dummies()
import pandas as pd

df = pd.DataFrame({'Color': ['Red', 'Blue', 'Green']})

encoded = pd.get_dummies(df, columns=['Color'])
print(encoded)

✔ Easy
❌ Not suitable for ML pipelines

✅ (B) Using scikit-learn (Recommended)
from sklearn.preprocessing import OneHotEncoder
import pandas as pd

df = pd.DataFrame({'Color': ['Red', 'Blue', 'Green']})

encoder = OneHotEncoder(sparse=False)

encoded = encoder.fit_transform(df[['Color']])

encoded_df = pd.DataFrame(
    encoded,
    columns=encoder.get_feature_names_out(['Color'])
)

print(encoded_df)

✔ Used in real ML projects
✔ Supports pipelines

🔹 6. Important Parameters (VERY IMPORTANT 🔥)
🔸 1. drop='first'
OneHotEncoder(drop='first')

👉 Removes one column to avoid redundancy

🔸 2. handle_unknown='ignore'
OneHotEncoder(handle_unknown='ignore')

👉 Prevents errors for unseen categories

🔹 7. Dummy Variable Trap
❗ What is it?

When all columns are present, one can be predicted from others → multicollinearity

👉 Example:
If we know:

Red = 0
Blue = 0

Then:

Green = 1 (automatically)
✅ Solution:
Use:
drop='first' (sklearn)
drop_first=True (pandas)
🔹 8. Advantages

✔ No false ordering
✔ Works well with most ML models
✔ Simple and interpretable

🔹 9. Disadvantages

❌ High dimensionality (too many columns)
❌ Not good for high-cardinality features (e.g., thousands of cities)

🔹 10. When to Use

✅ Use when:

Categories are small in number
No natural order exists

❌ Avoid when:

Too many unique categories
🔹 11. Real-Life Examples
Gender → Male, Female
Payment → Cash, Card, UPI
City → Delhi, Mumbai, Chennai
🔹 12. One Hot Encoding in ML Pipeline (Important)
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

ct = ColumnTransformer(
    transformers=[
        ('encoder', OneHotEncoder(), ['Color'])
    ],
    remainder='passthrough'
)

👉 Used in real-world ML projects

🔹 13. Common Mistakes

❌ Applying encoding separately on train & test
❌ Not handling unknown categories
❌ Using OHE for high-cardinality features

🔹 14. One-Line Summary

👉 One Hot Encoding converts categorical data into binary columns to avoid false relationships in ML models.
