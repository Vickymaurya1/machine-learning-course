**Implementing function transformer **


to use the dataset without removing the outliers.This method use when outliers contains some neccessary informantion foe the dataset

```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv("loan.csv")
dataset.head(3)
```

```python
dataset.isnull().sum()
```

```python
sns.distplot(dataset["CoapplicantIncome"])
plt.show()
```

```python
q1 = dataset['CoapplicantIncome'].quantile(0.25)
q3 = dataset["CoapplicantIncome"].quantile(0.75)
iqr = q3-q1
```

```python
min_range = q1 - (1.5*iqr)
max_range = q3+ (1.5*iqr)
min_range , max_range
```

```python
#dataset = dataset[dataset["CoapplicantIncome"] <= max_range]
```

```python
#implementing function transformer
from sklearn.preprocessing import FunctionTransformer
```

```python
ft = FunctionTransformer(func = np.log1p)
```

```python
ft.fit(dataset[["CoapplicantIncome"]])
```

```python
dataset["CoapplicantIncome_tf"] = ft.transform(dataset[["CoapplicantIncome"]])
```

```python
plt.figure(figsize = (10,4))
plt.subplot(1,2,1)
sns.displot(dataset["CoapplicantIncome"], kde = True)
plt.title("\nbefore")


plt.subplot(1,2,2)
plt.title("\nAfter")
sns.distplot(dataset["CoapplicantIncome_tf"], kde = True)
plt.show()
```
