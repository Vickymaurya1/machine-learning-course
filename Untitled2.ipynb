📘 ORDINAL ENCODING – FULL NOTES
🔹 1. Definition

Ordinal Encoding is used to convert categorical data with order into numerical values.

👉 Categories are encoded based on their ranking or importance

🔹 2. When to Use

✅ Use when data has natural order

🎯 Examples:
Size → Small < Medium < Large
Education → School < College < PhD
Rating → Low < Medium < High
🔹 3. Example (Animals by Size 🐾)
Animal Size
Small
Medium
Large
Medium

👉 Encoding:

Size	Value
Small	0
Medium	1
Large	2

✔ Order is meaningful → correct encoding

🔥 4. Method 1: Using sklearn

Using scikit-learn

📌 Code:
from sklearn.preprocessing import OrdinalEncoder
import pandas as pd

data = {'Size': ['Small', 'Medium', 'Large', 'Medium']}
df = pd.DataFrame(data)

encoder = OrdinalEncoder(categories=[['Small', 'Medium', 'Large']])

encoded = encoder.fit_transform(df[['Size']])

print(encoded)
✅ Output:
[[0.]
 [1.]
 [2.]
 [1.]]
🧠 Important

👉 You must define order manually:

categories=[['Small', 'Medium', 'Large']]
🔥 5. Method 2: Using map() function
📌 Code:
import pandas as pd

data = {'Size': ['Small', 'Medium', 'Large', 'Medium']}
df = pd.DataFrame(data)

mapping = {
    'Small': 0,
    'Medium': 1,
    'Large': 2
}

df['Size_encoded'] = df['Size'].map(mapping)

print(df)
✅ Output:
    Size  Size_encoded
0  Small       0
1  Medium      1
2  Large       2
3  Medium      1
🔹 6. Difference: OrdinalEncoder vs map()
