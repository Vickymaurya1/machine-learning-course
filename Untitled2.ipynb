Train–Test Split is a technique where we divide the dataset into two parts:

Training Data → Used to train the model

Testing Data → Used to evaluate the model

📊 Why Do We Split?

If we train and test on the same data, the model may memorize it (overfitting).
So we split to check how well the model performs on unseen data.

✅ Independent Variable (X)

Input variable

Features used to predict

Example: Age, Salary, Study Hours

👉 These influence the output.

✅ Dependent Variable (y)

Output variable

What we want to predict

Example: House Price, Exam Score, Disease Yes/No

👉 It depends on independent variables.

```python
import pandas as pd
```

```python
dataset = pd.read_csv("california_housing_test.csv")
dataset.head(3)
```

```python
input_data = dataset.iloc[: , :-1] #median_house_value will removed
output_data = dataset["median_house_value"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(input_data, output_data, test_size=0.25)
```
