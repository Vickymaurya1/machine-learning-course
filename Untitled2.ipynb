**Feature Selectio**n is the process of selecting the most important input features (variables) from a dataset to improve model performance and reduce complexity.

```python
import pandas as pd
from mlxtend.feature_selection import SequentialFeatureSelector
```

```python
dataset = pd.read_csv("diabetes.csv")
```

```python
dataset.head(3)
```

```python
x = dataset.iloc[: , :-1]
y = dataset["Outcome"]
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr = LogisticRegression()
```

```python
fs = SequentialFeatureSelector(lr, k_features= 3, forward = True)
fs.fit(x,y)
```

```python
fs.feature_names
```

```python
fs.k_feature_names_
```

```python
fs.k_score_
```
