OUTLIER DETECTION USING BOXPLOT
🔹 1. What is a Boxplot?

👉 A Boxplot (Box-and-Whisker plot) is a graph used to:

Show data distribution
Detect outliers visually
🔹 2. Structure of Boxplot
🧠 Key Parts:
Q1 (25%) → lower quartile
Median (Q2) → middle value
Q3 (75%) → upper quartile
Whiskers → range of normal data
Dots outside → 🚨 Outliers
🔹 3. How Boxplot Detects Outliers

👉 Uses IQR concept:

Lower limit = Q1 − 1.5 × IQR
Upper limit = Q3 + 1.5 × IQR

👉 Any point outside → Outlier

🔥 4. Example (Simple)

Data:

[10, 12, 15, 18, 100]

👉 Boxplot will show:

Most values inside box
100 as a separate dot → Outlier
🔹 5. Python Implementation
✅ Using matplotlib
import matplotlib.pyplot as plt

data = [10, 12, 15, 18, 100]

plt.boxplot(data)
plt.show()
✅ Using seaborn (Better Visualization)
import seaborn as sns

sns.boxplot(x=data)
🔹 6. Using Dataset Column
sns.boxplot(x=dataset["Salary"])
🔹 7. How to Interpret

👉 If you see:

Points outside whiskers → Outliers
Long whiskers → possible skewness
🔹 8. Multicolumn Example
sns.boxplot(data=dataset)

👉 Shows outliers for all columns

🔹 9. Advantages

✔ Easy to visualize
✔ Quick detection
✔ Works for large data
